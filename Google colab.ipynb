```python

causes = [
    "Habitudes anterieures de fraude",
    "Peur de l'echec et pression de la note",
    "Charge de travail",
    "Surveillance faible",
    "Technologies et IA",
    "Contexte socio-economique",
    "Influence des pairs",
    "Calcul risque-benefice",
    "Meconnaissance des regles"
]

probabilites = [84.62, 76.92, 69.23, 46.15, 53.85, 61.54, 69.23, 38.46, 30.77]
impacts = [60, 59.81, 50, 60, 46.54, 40, 26.73, 45, 35]

indices = []

for i in range(len(causes)):
    risque = probabilites[i] * impacts[i] / 100
    indices.append(risque)

print(indices)
```

```python

for i in range(len(causes)):
    print(causes[i], ":", round(indices[i], 2))
```

```python

classement = sorted(
    zip(indices, causes),
    reverse=True
)

print("CLASSEMENT DES RISQUES :")

for risque, cause in classement:
    print(cause, ":", round(risque, 2))
```
